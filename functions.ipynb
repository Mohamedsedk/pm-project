```python
def textAnalysis(text):
    def charCount(myText , char):
        return str(myText).count(char)
    
    
    def OriginalwordLengthDef(word):
        return len(word)
    
    def CleanedwordLengthDef(word):
        for char in '~!@#$%^&?':
            word = word.replace(char , '')
        return len(word)
    
    
    specialCharsDict = {}
    OriginalWordLengthDict = {}
    cleanedWordLengthDict = {}
    for i in '~!@#$%^&?':
        specialCharsDict[i] = charCount(text , i)
        
    for word in text.split():
        OriginalWordLengthDict[word] = OriginalwordLengthDef(word)
        
    for word in text.split():
        cleanedWordLengthDict[word]=CleanedwordLengthDef(word)
        
        
        
    return specialCharsDict , OriginalWordLengthDict , cleanedWordLengthDict
```

```python
phrase ='Hi!! how?? are~ you$$ '
textAnalysis(phrase)
```

```python
def textCleaning(text , special_char , limit = 100):
    text = text.lower()
    for char in special_char :
        text = text.replace(char , '')
    LimitedText = text[:limit]
    return LimitedText
```

```python
def fibon(n):
    a = 0 
    b = 1
    result= []
    for i in range(n):
        result.append(a)
        a,b = b , a+b 
    return result 
```

```python
print(fibon(7))
```
